```python
import pandas as pd
df = pd.read_csv("AG_News/test.csv")
df.head(5)
```

```python
class_map = {}

with open("AG_News/classes_name.txt", "r", encoding="utf-8") as f:
    for line in f:
        idx, label = line.strip().split(" ", 1)
        class_map[int(idx)] = label

print("Class mapping:", class_map)
```

```python
df["Class_Name"] = df["Class Index"].map(class_map)
df.head(5)
```
